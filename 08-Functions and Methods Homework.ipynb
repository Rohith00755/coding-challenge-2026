**Write a function that computes the volume of a sphere given its radius.**
<p>The volume of a sphere is given as $$\frac{4}{3} πr^3$$</p>

```python
def vol(rad):
    return (4/3)*(3.14)*(rad**3)
```

**Write a function that checks whether a number is in a given range (inclusive of high and low)**

```python
def ran_check(num,low,high):
    if num in range(low,high+1):
        print(f'{num} is in the range {low} and {high}')
    else:
        print(f'{num} is not in the range {low} and {high}')
```

```python
# Check
ran_check(8,2,7)
```

If you only wanted to return a boolean:

```python
def ran_bool(num,low,high):
    return True if num in range(low,high+1) else False
```

```python
ran_bool(11,1,10)
```

**Write a Python function that accepts a string and calculates the number of upper case letters and lower case letters.**

    Sample String : 'Hello Mr. Rogers, how are you this fine Tuesday?'
    Expected Output : 
    No. of Upper case characters : 4
    No. of Lower case Characters : 33

HINT: Two string methods that might prove useful: **.isupper()** and **.islower()**

If you feel ambitious, explore the Collections module to solve this problem!

```python
def up_low(s):
    u=0
    l=0
    for letter in s:
        if letter.isupper():
            u+=1
        if letter.islower():
            l+=1
    print(f'No. of Upper case characters : {u}')
    print(f'No. of Lower case Characters :  {l}')
```

```python
s = 'ello r. Rogers, how are you this fine Tuesday?'
up_low(s)
```

**Write a Python function that takes a list and returns a new list with unique elements of the first list.**

    Sample List : [1,1,1,1,2,2,3,3,3,3,4,5]
    Unique List : [1, 2, 3, 4, 5]

```python
def unique_list(lst):
    return list(set(lst))
    
```

```python
unique_list([1,1,1,1,2,2,3,3,3,3,4,5])
```

**Write a Python function to multiply all the numbers in a list.**

    Sample List : [1, 2, 3, -4]
    Expected Output : -24

```python
def multiply(numbers):  
    result=1
    for i in numbers:
        result *=i
    return result
        
```

```python
multiply([1,2,3,-4])
```

**Write a Python function that checks whether a word or phrase is palindrome or not.**

Note: A palindrome is word, phrase, or sequence that reads the same backward as forward, e.g., madam,kayak,racecar, or a phrase "nurses run". Hint: You may want to check out the .replace() method in a string to help out with dealing with spaces. Also google search how to reverse a string in Python, there are some clever ways to do it with slicing notation.

```python
def palindrome(s):
    s1=s[::-1]
    if s1 == s:
        return True
    else:
        return False
```

```python
palindrome('helleh')
```
